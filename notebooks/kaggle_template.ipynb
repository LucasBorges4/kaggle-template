# Kaggle + GitHub Template

Notebook pronto para clonar um repositório do GitHub e rodar treino/inferência no Kaggle.

1. Clone o repositório
2. Instale dependências
3. Treine o modelo
4. Gere o arquivo de submissão

```python
# CONFIGURE AQUI
GITHUB_USER = "SEU_USUARIO"
REPO_NAME = "kaggle-template"

import os
import subprocess

REPO_URL = f"https://github.com/{GITHUB_USER}/{REPO_NAME}.git"
DEST_DIR = f"/kaggle/working/{REPO_NAME}"

if not os.path.exists(DEST_DIR):
    print(f"Clonando {REPO_URL} ...")
    subprocess.run(["git", "clone", REPO_URL, DEST_DIR], check=True)
else:
    print(f"{DEST_DIR} já existe.")

import sys
if DEST_DIR not in sys.path:
    sys.path.insert(0, DEST_DIR)

print("Conteúdo:")
print(subprocess.check_output(["ls", "-la", DEST_DIR]).decode())
```
